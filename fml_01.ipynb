### Quick CheatSheet

| Action                  | `loc` – By Label                        | `iloc` – By Position    |
| ----------------------- | --------------------------------------- | ----------------------- |
| Select a single row     | `data.loc[2]`                           | `data.iloc[2]`          |
| Select a single column  | `data.loc[:, "Temperature"]`            | `data.iloc[:, 3]`       |
| Select multiple rows    | `data.loc[[1, 3, 5]]`                   | `data.iloc[[1, 3, 5]]`  |
| Select multiple columns | `data.loc[:, ["Date", "Temperature"]]`  | `data.iloc[:, [0, 3]]`  |
| Select a range of rows  | `data.loc[:5]`                          | `data.iloc[:5]`         |
| Select rows and columns | `data.loc[:5, ["Date", "Temperature"]]` | `data.iloc[:5, [0, 3]]` |
| Select from the end     | —                                       | `data.iloc[-5:, -2:]`   |

### Cheat Sheet: Comparison Operators  

| Operator | Meaning                  |
| -------- | ------------------------ |
| `==`     | Equal to                 |
| `!=`     | Not equal to             |
| `>`      | Greater than             |
| `<`      | Less than                |
| `>=`     | Greater than or equal to |
| `<=`     | Less than or equal to    |

🧩 Let's select only the observations from summer using a Boolean mask.

```python
mask = df["Seasons"] == "Summer"
df[mask]
```

🧩  Let's select all observations that are not from summer.

```python
mask = df["Seasons"] != "Summer"
df[mask]
```

```python
df[df["Seasons"] != "Summer"]
```

#### Combining Multiple Conditions  

| Operator | Meaning |
| -------- | ------- |
| `&`      | AND     |
| `\|`     | OR      |
| `~`      | NOT     |

🧩 Let's select observations where the temperature is above 25 °C and the humidity is below 60%.

```python
mask = (df["Temperature"] > 25) & (df["Humidity"] < 60)
df[mask]
```

🧩 Let's select observations where the temperature is above 25 °C or the humidity is above 80%.

```python
mask = (df["Temperature"] > 25) & (df["Humidity"] > 80)
df[mask]
```

💡 When combining multiple conditions, put each condition in parentheses ().

🧩 Let's select observations that are not from summer.

```python
mask = ~(df["Seasons"] == "Summer")
df[mask]
```

💡 The ~ operator negates a Boolean condition: True becomes False and False becomes True.

- #### **Grouping** is the process of splitting data into groups based on the values of one or more columns.  
- #### **Aggregation** is the process of computing a summary statistic for each group.

🧩 Let's calculate the average number of rented bikes for each season.

```python
df.groupby("Seasons")["Rented_bike_count"].mean()
```

🧩 Let's calculate the median number of rented bikes for each season and sort the results.

```python
df.groupby("Seasons")["Rented_bike_count"].median().sort_values()
```

💡 By default, sort_values() sorts values in ascending order.

```python
df.groupby("Seasons")["Rented_bike_count"].median().sort_values(ascending=False)
```

🧩 Let's calculate the average number of rented bikes for each hour of the day.

```python
df.groupby("Hour")["Rented_bike_count"].mean()
```

### Matplotlib

```python
import matplotlib.pyplot as plt
```

- #### Line plot

🧩 Let's visualize the average number of rented bikes for each hour of the day.

```python
hourly_demand = df.groupby("Hour")["Rented_bike_count"].mean()
plt.figure(figsize=(10, 5))
plt.plot(hourly_demand.index, hourly_demand.values, marker="o")

plt.xlabel("Hour")
plt.ylabel("Average rented bike count")
plt.title("Average Bike Demand During the Day")

plt.show()
```

By default, Matplotlib connects the observations with a line.

We can plot multiple lines in the same figure. Each plot() call adds another line to the current plot.

```python
hourly_mean = df.groupby("Hour")["Rented_bike_count"].mean()
hourly_max = df.groupby("Hour")["Rented_bike_count"].max()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(hourly_mean.index, hourly_mean.values, label="Mean",marker="o")

plt.plot(hourly_max.index, hourly_max.values,label="Maximum", marker="^")

plt.xlabel("Hour")
plt.ylabel("Rented bike count")
plt.title("Bike Demand During the Day")
plt.legend()

plt.show()
```

### Interactive Plotly chart

Plotly creates interactive charts: you can hover over points, zoom, pan, and hide categories from the legend. We will use the high-level `plotly.express` interface.

```python
import plotly.express as px
```

```python
fig = px.scatter(df,
    x="Temperature",
    y="Rented_bike_count",
    color="Seasons",
    hover_data=["Hour"],
    title="Bike rentals by temperature and season",
    labels={"Temperature": "Temperature (°C)",
        "Rented_bike_count": "Rented bike count",
        "Seasons": "Season",},
    opacity=0.65,)

fig.show(renderer="browser")
```
